# Structure

The project consists of 2 parts:
* core/ - the core part of an agent(Pydentic models and clean logic). This package is distinct from the WWW or LLM.

* connectors/ - "hands" and "eyes" of the agent (Dou access, Gemini call, sending letters)

Folders:
helpers/ - auxiliary code (e.g. compound HTTP query to DOU with cookies)
data/ - storage of your CV in PDF format
fixtures/ - cached data
